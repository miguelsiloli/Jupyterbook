## Introduction

This Exploratory Data Analysis (EDA) delves into the narrative of Portugal's contemporary housing market. It's partly a **'Tale of Two Cities,'** where Lisbon and Porto, along with their greater metropolitan areas (including contiguous councils like Setúbal, Cascais, Santarém for Lisbon, and Gaia with its satellite cities for Porto), stand as bustling hubs of intense activity and soaring property values. This dynamic often overshadows the realities faced by the rest of the nation, creating distinct and often disparate cultural and housing landscapes.

The source of the data is imovirtual listings.

---

## The Stark Price Contrast

The data unequivocally points to a housing market operating under **two vastly different price realities**. A significant premium is demanded to live within or near the metropolitan areas of Lisbon and Porto, a fact starkly illustrated by Kernel Density Estimate (KDE) plots and comparative price tables from our analysis.

Furthermore, the concentration of high-value outliers within these metropolises strongly suggests a **vibrant luxury market** that is largely absent in other regions. Lisbon and Porto are not just population centers; they are **trendy**, high-demand cities on the international stage.

### Socioeconomic Implications: Real Estate Appreciation Doesn't Necessarily Create Broad Value

The allure of high returns in Lisbon and Porto funnels significant capital, both domestic and international, into these established markets. This capital, which could be vital for fostering high-specialization jobs and diversifying the economy, is instead predominantly channeled into:
*   Real estate development and speculation.
*   The tourism sector and its associated infrastructure (e.g., Airbnbs).
*   Jobs that, while providing employment, often offer lower value-added contributions to the broader economy.

> **The "Trickle-Down" Fallacy:** Contrary to trickle-down economic theories, this concentration of wealth and investment in real estate appears to primarily benefit a smaller segment of the population—those owning these businesses, properties for rent, or involved in high-end development. **Real estate appreciation is a mechanism of generational wealth transfer**

*   **Specialized Workforce Pushed Away:** The dual pressures of a lack of investment in specialized, high-skilled sectors *within* these cities, combined with the intense strain on basic services like transport and the exorbitant cost of housing, are actively pushing valuable, skilled individuals out of these major urban centers, or even out of the country entirely.

---

Beyond just price, the **sheer volume of listings** (4448 in Lisbon & Porto vs. 1716 across the 'Rest of Portugal' in our dataset) underscores the overwhelming dominance of these two urban centers in the national housing market.

## A New Meaning of Home: From a Family backbone to Financial Asset

The overwhelming prevalence of **'business' or 'professional/institutional entities' as advertisers** (accounting for 5791 listings versus only 372 for 'private' advertisers in our data) signals a fundamental, albeit global, shift. Housing in Portugal is increasingly perceived and transacted not as a cornerstone of social welfare with intrinsic sentimental value, but as a **financial asset and an economic service.**

**The houses is now fundamentally marketed as a financial possession and not the safe haven of families**.

### Questions and Consequences Arising from This Shift:

*   If the primary actors in the housing market are large, often impersonal, businesses, **does this erode the sense of community and the social contract** that once underpinned neighborhoods?
*   Are current **regulatory frameworks equipped to manage a housing market where large institutional players are so dominant?**
*   When homes are primarily treated as investment vehicles and decisions are driven by profit maximization, **who will advocate for the housing needs and well-being of local populations?**

## The Unaffordable Reality: Wages vs. Walls

The national median gross monthly wage in Portugal stands at approximately **€1,014**, while the statutory minimum gross monthly wage is around **€820** (as of early 2024). These figures starkly contrast with the cost of renting in the major cities:

**Lisbon:**
*   Even a **T0 (studio apartment)** commands a median rent of **€980 per month**. This alone is alarmingly close to the entire gross median monthly wage and significantly exceeds the gross minimum wage.
*   A **T1 (one-bedroom apartment)** sees median rent climb to **€1,350**, substantially more than the national median gross wage.
*   A **T2 (two-bedroom apartment)**, often the minimum for a small family, reaches a median of **€1,650** – far outstripping what an individual on a median wage could afford, necessitating multiple high incomes per household.
*   Larger apartments like **T3s (€2,200)** and **T4s (€3,000)** are firmly luxury rentals.
*   (The outlier T9 at €1,300 based on a single listing is likely an anomaly.)

**Porto:**
*   While generally less expensive, Porto presents a significant challenge. A **T0 median rent is €850**.
*   A **T1 at €998** is nearly identical to the national median gross wage.
*   A **T2 at €1,250** and a **T3 at €1,650** quickly outpace average earning capacity.
*   (The T5 at €2,000 and T9+ at €7,500 highlight specialized/luxury segments.)

### Living on the Edge

Considering the financial guideline that housing costs should not exceed **30-40% of net income**:

*   **Minimum Wage Earner (€820 gross):** It is virtually impossible to independently afford *any* listed typology in either Lisbon or Porto without falling into extreme housing stress (rent consuming well over 50-100%+ of gross income). This forces co-habitation, often in overcrowded conditions, reliance on limited social housing, or prohibitive commutes.
*   **Median Wage Earner (€1,014 gross):**
    *   In Lisbon, a T0 (€980) would consume approximately **97% of their gross monthly income**.
    *   In Porto, a T0 (€850) would take up about **84% of their gross income**; a T1 (€998) is essentially their entire gross median wage.
    *   This forces difficult choices: roommates, long commutes, or sacrificing other essential needs.

## Real Estate and Tourism: Portugal's New Dutch Disease?

The "Dutch Disease" describes how rapid development in one sector can harm others. In Portugal, real estate and tourism may be playing this role.

Factors like **Golden Visa programs, foreign investment, a surge in short-term rentals (e.g., Airbnb), and Portugal's desirability** have caused property values in key areas (Lisbon, Porto, Algarve) to skyrocket. Simultaneously, Portugal's status as a **top global tourist destination**, while bringing revenue, places immense pressure on local infrastructure, especially housing.

### Crowding Out of Other Economic Activities:

*   **Local Businesses:** Traditional businesses in gentrifying areas are often forced out by soaring commercial rents, replaced by those catering primarily to tourists or high-income residents. This leads to a loss of local economic diversity and character.
*   **Manufacturing/Tradable Sectors:** Though not a heavy manufacturing economy, Portugal's other export-oriented businesses face higher operational costs (rent, potential labor inflation) without directly benefiting from the real estate/tourism boom, eroding their competitiveness.
*   **Innovation & Diversification:** An economy overly focused on real estate returns and tourism services may underinvest in long-term innovation. The "easy money" in property can disincentivize riskier, but potentially more transformative, ventures in other fields.
*   **Vulnerability:** Over-reliance on tourism makes the economy highly vulnerable to external shocks, as evidenced by the COVID-19 pandemic's impact on the sector.

---

## Synopsis: The Human Cost of a Market Imbalance

The current state of Portugal's housing market, particularly in its major cities, has profound and often detrimental impacts on its population:

*   **Forced Co-habitation & Overcrowding:** Individuals and families are increasingly pushed into shared living arrangements out of necessity, impacting privacy, well-being, and public health, alongside a general erosion of quality of life.
*   **Exclusion of Essential Workers:** Key workers like nurses, teachers, and service staff, whose salaries often hover around the median wage, are priced out of the cities they serve. This leads to labor shortages in critical sectors or forces these workers into extremely long and costly commutes.
    > *A concerning potential "solution" often floated is to rely on increased immigration to fill these essential worker gaps at low pay, further exacerbating issues of overcrowded and precarious living conditions for these new arrivals.*
*   **Increased Social Inequality:** The housing market bifurcates society. Those with capital or very high incomes access central, well-serviced housing, while a large segment is pushed to the margins, both geographically and economically.
*   **Youth Displacement & "Generation Rent":** Young people entering the workforce find independence nearly impossible. The prospect of saving for a down payment is a distant dream when monthly rents are cripplingly high, trapping them in a "generation rent" cycle with little opportunity to build equity or financial stability.
*   **Impact on Family Formation:** The financial strain and lack of suitable, affordable housing can lead individuals and couples to delay starting families or to have fewer children than they might desire. If people cannot secure a stable home, how can they realistically plan for a family?

---

In short, the housing market reflects a mixture of global transitions of houses as a financial asset and the result of the political landscape in Europe.

It has become an extreme case study in how market forces, without sufficient social safeguards and strategic long-term planning, can lead to **cultural inequality and a sense of expropriation for a significant portion of its own citizens.**

```python
# Set a nicer global style for plots
sns.set_theme(style="whitegrid", palette="pastel")

MAX_COMMON_ITEMS_TO_DISPLAY = 10 # For the table of exploded features

for col_name in ['features', 'featuresByCategory', 'featuresWithoutCategory']:
    col_title = col_name.replace('_', ' ').title()
    display(HTML(f"<h2>Analysis of List Column: {col_title}</h2>"))

    # --- 1. Parse the List-like Column ---
    def parse_feature_list_revised(val): # Keep your robust parser
        if isinstance(val, list): return val
        if isinstance(val, str):
            try:
                parsed_val = ast.literal_eval(val)
                return parsed_val if isinstance(parsed_val, list) else []
            except (ValueError, SyntaxError): return []
        if val is None or (pd.api.types.is_scalar(val) and pd.isna(val)): return []
        if isinstance(val, np.ndarray): return val.tolist()
        return []

    df[f'{col_name}_list'] = df[col_name].apply(parse_feature_list_revised)
    df[f'num_{col_name}'] = df[f'{col_name}_list'].apply(len)
    assert df[f'{col_name}_list'].apply(lambda x: isinstance(x, list)).all(), f"Column {col_name}_list not all lists!"

    # --- 2. Display Sample Data and Parsed Lists (Styled HTML Table) ---
    display(HTML("<h4>Sample Data & Parsed Lists (First 5 Rows):</h4>"))
    sample_df_display = df[[col_name, f'{col_name}_list', f'num_{col_name}']].head().copy()
    # Truncate long list representations for display
    sample_df_display[f'{col_name}_list'] = sample_df_display[f'{col_name}_list'].apply(lambda x: str(x)[:70] + '...' if len(str(x)) > 70 else str(x))
    sample_df_display.columns = ['Original Value', 'Parsed List (Sample)', 'Num Items']
    display(sample_df_display.style.set_caption("Sample Data").set_properties(**{'text-align': 'left', 'max-width': '300px'}))


    # --- 3. Display Descriptive Stats for Number of Items (Styled HTML Table) ---
    display(HTML("<h4>Descriptive Statistics for Number of Items per Listing:</h4>"))
    desc_stats_num_items = df[f'num_{col_name}'].describe().to_frame().round(2)
    desc_stats_num_items.columns = ['Value']
    display(desc_stats_num_items.style.set_caption(f"Stats for num_{col_name}"))


    # --- 4. Plot Histogram for Number of Items ---
    if df[f'num_{col_name}'].notna().any() and df[f'num_{col_name}'].nunique() > 0:
        plt.figure(figsize=(10, 5)) # Adjusted size for single plot
        sns.histplot(df[f'num_{col_name}'],
                     bins=max(1, min(20, df[f'num_{col_name}'].nunique())),
                     kde=False, color="skyblue")
        plt.title(f'Distribution of Number of Items in {col_title}', fontsize=14)
        plt.xlabel('Number of Items', fontsize=12)
        plt.ylabel('Frequency of Listings', fontsize=12)
        plt.tight_layout()
        plt.show()
    else:
        display(HTML("<p><em>No data or no variance in the number of items to plot.</em></p>"))


    # --- 5. Analyze and Display Top Common Exploded Items (Styled HTML Table & Optional Bar Plot) ---
    first_item_type = None
    can_explode_simple_features = False
    top_common_items_series = None

    if df[f'num_{col_name}'].sum() > 0: # Only check if there are any items at all
        for item_list in df[f'{col_name}_list']:
            if item_list: # If list is not empty
                actual_item = next((item for item in item_list if not (isinstance(item, float) and np.isnan(item))), None) # Get first non-NaN item
                if actual_item is not None:
                    if isinstance(actual_item, (str, int, float, bool)):
                        first_item_type = type(actual_item)
                        can_explode_simple_features = True
                    else:
                        first_item_type = 'complex'
                        display(HTML(f"<p><em>Note: Items in '{col_title}' lists are complex (e.g., dicts like: {str(actual_item)[:100]}...). Skipping common item analysis.</em></p>"))
                    break # Found type, no need to check further lists

    if can_explode_simple_features:
        display(HTML(f"<h4>Top {MAX_COMMON_ITEMS_TO_DISPLAY} Most Common Individual Items in '{col_title}':</h4>"))
        try:
            series_to_explode = df[df[f'num_{col_name}'] > 0][f'{col_name}_list']
            if not series_to_explode.empty:
                exploded_features = series_to_explode.explode().dropna() # Explode and drop NaNs from exploded series
                if not exploded_features.empty:
                    top_common_items_series = exploded_features.value_counts(dropna=True).nlargest(MAX_COMMON_ITEMS_TO_DISPLAY)

                    if not top_common_items_series.empty:
                        top_common_items_df = top_common_items_series.reset_index()
                        top_common_items_df.columns = ['Item', 'Frequency']
                        top_common_items_df['Percentage (%)'] = (top_common_items_df['Frequency'] / len(exploded_features) * 100).round(2)
                        display(top_common_items_df.style.set_caption(f"Top Common Items in {col_title}")
                                                        .bar(subset=['Percentage (%)'], color='#5dade2', vmin=0)
                                                        .format({'Percentage (%)': '{:.2f}%'}))

                        # Optional: Bar plot for these top common items
                        plt.figure(figsize=(10, max(5, len(top_common_items_series) * 0.4)))
                        sns.barplot(x='Frequency', y='Item', data=top_common_items_df, palette="viridis", dodge=False) # dodge=False if only one series
                        plt.title(f'Top {len(top_common_items_series)} Common Items in {col_title}', fontsize=14)
                        plt.xlabel('Frequency', fontsize=12)
                        plt.ylabel('Item', fontsize=12)
                        # Add counts to bars
                        for index, row in top_common_items_df.iterrows():
                            plt.text(row.Frequency + (plt.xlim()[1]*0.005), index, str(int(row.Frequency)), color='black', va='center', fontsize=9)
                        current_xlim = plt.xlim()
                        plt.xlim(current_xlim[0], current_xlim[1]*1.05)
                        plt.tight_layout()
                        plt.show()
                    else:
                        display(HTML("<p><em>No common items found after exploding and dropping NaNs.</em></p>"))
                else:
                    display(HTML("<p><em>No non-NaN items to analyze after exploding the lists.</em></p>"))
            else:
                display(HTML(f"<p><em>No lists with items to explode in '{col_title}'.</em></p>"))
        except Exception as e:
            display(HTML(f"<p><em>Could not explode or process items for '{col_title}'. Error: {e}.</em></p>"))

    display(HTML("<hr>")) # Separator after each column's analysis
```

```python
# --- Calculate Median Prices by Typology for Lisbon and Porto ---
if 'address_district' in df_lp.columns and 'typology' in df_lp.columns and 'price' in df_lp.columns:
    # Filter for Lisbon and Porto explicitly if df_lp contains more than just these two
    # (though df_lp should already be filtered)
    df_lisbon = df_lp[df_lp['address_district'] == 'Lisboa'].copy()
    df_porto = df_lp[df_lp['address_district'] == 'Porto'].copy()

    # Calculate median price - median is often more robust to outliers than mean for prices
    median_price_lisbon_typology = df_lisbon.groupby('typology')['price'].median().round(0)
    median_price_porto_typology = df_porto.groupby('typology')['price'].median().round(0)

    # Count of listings for context
    count_lisbon_typology = df_lisbon.groupby('typology')['price'].count()
    count_porto_typology = df_porto.groupby('typology')['price'].count()

    # Combine into a single DataFrame for display
    typology_price_comparison = pd.DataFrame({
        'Typology': median_price_lisbon_typology.index,
        'Lisbon Median Price (€)': median_price_lisbon_typology.values,
        'Lisbon Listings Count': count_lisbon_typology.reindex(median_price_lisbon_typology.index).values
    }).set_index('Typology')

    porto_df_temp = pd.DataFrame({
        'Typology': median_price_porto_typology.index,
        'Porto Median Price (€)': median_price_porto_typology.values,
        'Porto Listings Count': count_porto_typology.reindex(median_price_porto_typology.index).values
    }).set_index('Typology')

    typology_price_comparison = typology_price_comparison.join(porto_df_temp, how='outer').fillna(0) # Fill NaN with 0 for counts/prices if a typology is missing in one city

    # Ensure integer types for display where appropriate
    for col in ['Lisbon Median Price (€)', 'Porto Median Price (€)', 'Lisbon Listings Count', 'Porto Listings Count']:
        if col in typology_price_comparison.columns:
            typology_price_comparison[col] = typology_price_comparison[col].astype(int)


    # Sort by a typical typology order if possible (T0, T1, T2...)
    # Create a custom sort order for typologies
    typology_order = [f'T{i}' for i in range(10)] + [f'T{i}+' for i in range(10)] + ['T10+','Estúdio', 'Outro'] # Add more common typologies if needed
    # Convert index to Categorical with the defined order
    typology_price_comparison.index = pd.Categorical(typology_price_comparison.index, categories=typology_order, ordered=True)
    typology_price_comparison = typology_price_comparison.sort_index()

    # Reset index to make 'Typology' a column again for styling
    typology_price_comparison.reset_index(inplace=True)


    display(HTML("<h3>Median Housing Prices by Typology: Lisbon vs. Porto</h3>"))
    # Style the DataFrame
    styled_table = typology_price_comparison.style.set_caption("Median Property Prices (€) and Listing Counts by Typology") \
        .format({
            'Lisbon Median Price (€)': "€{:,.0f}",
            'Porto Median Price (€)': "€{:,.0f}",
            'Lisbon Listings Count': "{:,.0f}",
            'Porto Listings Count': "{:,.0f}"
        }) \
        .set_properties(**{'text-align': 'right', 'min-width': '100px'}) \
        .set_table_styles([
            {'selector': 'th', 'props': [('text-align', 'center'), ('font-weight', 'bold')]},
            {'selector': 'td, th', 'props': [('border', '1px solid #ddd')]},
            {'selector': 'caption', 'props': [('caption-side', 'top'), ('font-size', '1.1em'), ('font-weight', 'bold'), ('margin-bottom', '10px')]}
        ]) \
        .bar(subset=['Lisbon Median Price (€)', 'Porto Median Price (€)'], color='#5dade2', vmin=0, align='zero') \
        .hide(axis="index") # Hide the default numerical index

    display(styled_table)

else:
    display(HTML("<p style='color:orange;'>Warning: Could not generate typology price table. Required columns ('address_district', 'typology', 'price') not found in df_lp or df_lp is empty.</p>"))
```
